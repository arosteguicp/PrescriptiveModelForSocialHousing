```python
import pandas as pd

df = pd.read_csv("/content/osm_distancias_long_checkpoint (7).csv")

translation = {
    "Bus": "Bus Stops",
    "Alimentacion": "Food",
    "Farmacias": "Pharmacies",
    "Tienda": "Stores",
    "Colegio Primaria": "Primary Schools",
    "Colegio Secundaria": "Secondary Schools",
    "Hospital": "Hospitals",
    "Parque": "Parks",
    "Banco": "Banks",
    "Mercado": "Markets"
}

df["category_en"] = df["categoria"].replace(translation)

df["category_en"] = df["category_en"].fillna(df["categoria"])

# ==========================
# Conteo por proyecto
# ==========================
counts = (
    df.groupby(["category_en", "id_proyecto"])
      .size()
      .reset_index(name="count")
)

count_stats = (
    counts.groupby("category_en")["count"]
    .agg(
        Total_Count="sum",
        Average_Count="mean",
        Min_Count="min",
        Max_Count="max"
    )
)

# ==========================
# Estadísticas de distancia
# ==========================
distance_stats = (
    df.groupby("category_en")["dist_m"]
      .agg(
          Average_Distance_m="mean",
          Distance_Std_m="std"
      )
)

# ==========================
# Tabla final
# ==========================
summary = (
    count_stats
    .join(distance_stats)
    .round(2)
    .sort_index()
)

summary = summary.reset_index().rename(columns={
    "category_en": "Category"
})

print(summary)

# Si deseas guardarla
summary.to_csv("category_statistics.csv", index=False)

print("\nArchivo guardado como category_statistics.csv")
```

```python
from PIL import Image, ImageDraw, ImageOps

def crop_to_circle(path_in, path_out, size=600):
    img = Image.open(path_in).convert("RGBA")
    img = ImageOps.fit(img, (size, size), Image.LANCZOS)  # recorta a cuadrado
    mask = Image.new("L", (size, size), 0)
    draw = ImageDraw.Draw(mask)
    draw.ellipse((0, 0, size, size), fill=255)
    img.putalpha(mask)
    img.save(path_out, "PNG")

crop_to_circle("AlamedaRimac0_181.png", "AlamedaRimac0_181_circle.png")
crop_to_circle("AlamedaRimac_Counterfactual.png", "AlamedaRimac_Counterfactual_circle.png")
crop_to_circle("AlamedaRimac_POIdist.png", "AlamedaRimac_POIdist_circle.png")
```
